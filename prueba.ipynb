buenas tardes (oscar)

Las variables que más influyen en el precio de un coche son también las que el comprador consulta para buscar un coche. En la extracción de características pudimos observar que los factores con mayor peso en la determinación del precio son la potencia del motor, el par motor y el kilometraje. Es frecuente que los compradores busquen un coche que tenga poco desgaste acumulado, por ello se recomienda reducir el precio de coches con alto kilometraje para acelerar su venta. Respecto a la potencia del motor, los coches con mayor potencia suelen ser superdeportivos donde los clientes interesados están dispuestos a comprarlos a un precio alto, por lo que se recomienda aprovechar al margen la venta de estos coches. En cuanto a la fuerza de tracción del motor, destacan las furgonetas y pick-ups destinados al transporte de objetos. Se recomienda ajustar el precio de estos vehículos según sus cualidades físicas: si el vehículo tuviese un par motor elevado, se valoraría mejor su tamaño y tipo de carrocería que el de un coche con un par motor bajo.
